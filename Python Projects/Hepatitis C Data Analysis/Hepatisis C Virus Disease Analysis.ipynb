### Biochemical Markers

#### ALB (Albumin)

Albumin is a protein produced by the liver.

Importance:
Low albumin levels may indicate poor liver function or chronic liver disease.

Use:
- Detect liver damage
- Compare healthy vs diseased individuals

#### ALP (Alkaline Phosphatase)

ALP is an enzyme related to bile ducts.

Importance:
High ALP levels may indicate bile duct obstruction or liver disease.

Use:
- Identify liver and bone disorders

#### ALT (Alanine Aminotransferase)

ALT is a key liver enzyme.

Importance:
Elevated ALT levels strongly indicate liver damage.

Use:
- One of the most important indicators of liver disease
- Used in risk score calculation

#### AST (Aspartate Aminotransferase)

AST is another liver enzyme.

Importance:
High AST levels indicate liver injury or inflammation.

Use:
- Often analyzed along with ALT
- Helps assess severity of liver damage

#### BIL (Bilirubin)

Bilirubin is a waste product processed by the liver.

Importance:
High bilirubin levels cause jaundice and indicate liver dysfunction.

Use:
- Detect advanced liver disease (e.g., cirrhosis)

#### CHE (Cholinesterase)

CHE is an enzyme produced by the liver.

Importance:
Low levels may indicate liver damage.

Use:
- Assess liver function

#### CHOL (Cholesterol)

Cholesterol levels reflect metabolic and liver health.

Importance:
Liver plays a role in cholesterol metabolism.

Use:
- Study relationship between metabolism and liver disease

#### CREA (Creatinine)

Creatinine is related to kidney function.

Importance:
Although not directly a liver marker, it helps assess overall organ health.

Use:
- Identify multi-organ impact

#### GGT (Gamma Glutamyl Transferase)

GGT is an enzyme linked to liver and bile duct function.

Importance:
High GGT levels strongly indicate liver disease or alcohol-related damage.

Use:
- Important for detecting liver dysfunction

#### PROT (Total Protein)

Total protein measures overall protein levels in blood.

Importance:
Reflects liver’s ability to synthesize proteins.

Use:
- Assess overall liver function

## TASK 2: Data Cleaning & Pre-processing

```python
df.drop(columns=["Unnamed: 0"], inplace=True)
```

The column 'Unnamed: 0' is an index-like column that does not provide any meaningful information for analysis. It was likely created during data export. Keeping such columns can introduce noise and affect model performance, so it is removed.

```python
df["Category"] = df["Category"].str.split("=").str[1]
```

The 'Category' column contains combined values such as "0=Blood Donor". This format is not suitable for analysis or modeling. Therefore, we extract only the meaningful label (e.g., "Blood Donor") by splitting the string.

```python
df["Sex"] = df["Sex"].map({"m":1, "f":0})
```

The 'Sex' column contains categorical values ('m' and 'f'), which cannot be directly used in numerical analysis or machine learning models. Therefore, it is converted into numerical form using encoding (male = 1, female = 0).

```python
df.isnull().sum()
```

```python
df.fillna(df.median(numeric_only=True), inplace=True)
```

The dataset contains missing values in several numerical columns. These missing values can negatively impact analysis and model performance.

Median imputation is used instead of mean because:
- It is more robust to outliers
- It preserves the distribution of skewed medical data

This ensures that missing values are handled without introducing bias.

```python
df.drop_duplicates(inplace=True)
```

Duplicate rows can distort analysis by over-representing certain observations. Removing duplicates ensures that each data point is unique and prevents bias in statistical analysis and modeling.

```python
df["Age"] = df["Age"].astype(int)
```

#### Average Biomarkers by Disease Category

```python
df.groupby("Category")[["ALT","AST","BIL","ALB"]].mean()
```

#### Created new column Liver_Risk_Score

```python
df["Liver_Risk_Score"] = df[["ALT","AST","BIL"]].mean(axis=1)
df.head()
```

#### Average Age by Category

```python
df.groupby("Category")["Age"].mean()
```

## TASK 3: EDA & Visualizations

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

### Age Distribution

```python
plt.figure(figsize=(12,6))
sns.histplot(df["Age"], kde=True)
plt.title("Age Distribution")
plt.show()
```

This histogram shows the distribution of age across all individuals in the dataset.

We can observe that most individuals fall within the middle age range, indicating that liver-related conditions are more commonly observed in this age group.

The distribution appears slightly skewed, suggesting that fewer very young or very old individuals are present in the dataset.

This helps us understand the demographic concentration and potential age-related risk patterns.

### Category Distribution

```python
plt.figure(figsize=(12,6))
sns.countplot(x="Category", data=df)
plt.title("Category Distribution")
plt.show()
```

This plot shows the distribution of individuals across different health categories such as Blood Donors, Hepatitis, Fibrosis, and Cirrhosis.

We can clearly see that the majority of individuals are healthy blood donors, while fewer cases belong to disease categories.

This imbalance is important because it indicates that the dataset is skewed, which can affect model training and interpretation.

### Age vs Disease Category

```python
plt.figure(figsize=(12,6))
sns.boxplot(x="Category", y="Age", data=df)
plt.title("Age vs Disease Category")
plt.show()
```

This box plot compares the age distribution across different disease categories.

We can observe differences in median age and spread across categories, suggesting that certain diseases are more prevalent in specific age groups.

For example, more severe conditions like Cirrhosis may be associated with older individuals.

This helps identify age as a potential risk factor.

### ALT vs AST

```python
sns.scatterplot(x="ALT", y="AST", hue="Category", data=df)
plt.title("ALT vs AST")
plt.show()
```

This scatter plot shows the relationship between ALT and AST, which are key liver enzymes.

A strong positive relationship can be observed, meaning that as ALT increases, AST also tends to increase.

Additionally, different categories are clustered in distinct regions, indicating that these biomarkers are effective in distinguishing between healthy and diseased individuals.

This is a strong indicator that these features are important for classification.

## Exploratory Data Analysis (EDA)

## Scatter Plot: Acquisition Cost vs ROI

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Acquisition_Cost', y='ROI', data=df)
plt.title("Acquisition Cost vs ROI")
plt.show()
```

##  Bar Chart - Average Conversion Rate by Channel and Campaign Type

```python
plt.figure(figsize=(10,6))
sns.barplot(x='Channel_Used', y='Conversion_Rate', hue='Campaign_Type', data=df)
plt.title("Average Conversion Rate by Channel and Campaign Type")
plt.show()
```

## Box Plot - Engagement Score by Campaign Type

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='Campaign_Type', y='Engagement_Score', data=df)
plt.title("Engagement Score Distribution by Campaign Type")
plt.show()
```

## Bar Chart - Average ROI by Company

```python
roi_company = df.groupby('Company')['ROI'].mean().reset_index()

plt.figure(figsize=(10,6))
sns.barplot(x='Company', y='ROI', data=roi_company)
plt.xticks(rotation=45)
plt.title("Average ROI by Company")
plt.show()
```

## Heatmap - Correlation Heatmap

```python
corr = df[['Engagement_Score','Conversion_Rate']].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

## Customer Segmentation

### Count Plot - Target Audience Distribution

```python
plt.figure(figsize=(10,6))
sns.countplot(x='Target_Audience', data=df)
plt.xticks(rotation=45)
plt.title("Distribution of Target Audience")
plt.show()
```

### Bar Chart - Conversion Rate by Customer Segment and Language

```python
plt.figure(figsize=(10,6))
sns.barplot(x='Customer_Segment', y='Conversion_Rate', hue='Language', data=df)
plt.xticks(rotation=45)
plt.title("Conversion Rate by Customer Segment and Language")
plt.show()
```

### Box Plot - Acquisition Cost Distribution by Customer Segment and Channel

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='Customer_Segment', y='Acquisition_Cost', hue='Channel_Used', data=df)
plt.xticks(rotation=45)
plt.title("Acquisition Cost by Customer Segment and Channel")
plt.show()
```

### Bar Chart - Average Conversion Rate by Language

```python
conv_lang = df.groupby('Language')['Conversion_Rate'].mean().reset_index()

plt.figure(figsize=(6,4))
sns.barplot(x='Language', y='Conversion_Rate', data=conv_lang)
plt.title("Average Conversion Rate by Language")
plt.show()
```

## Channel Effectiveness

### Engagement Score by Channel and Campaign Type

```python
plt.figure(figsize=(10,6))
sns.barplot(x='Channel_Used', y='Engagement_Score', hue='Campaign_Type', data=df)
plt.title("Engagement Score by Channel and Campaign Type")
plt.show()
```

### Pie Chart ROI Distribution by Channel

```python
roi_channel = df.groupby('Channel_Used')['ROI'].sum()

plt.figure(figsize=(6,6))
plt.pie(roi_channel, labels=roi_channel.index, autopct='%1.1f%%')
plt.title("ROI Distribution by Channel")
plt.show()
```

### Scatter Plot - Clicks vs Impressions

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='Impressions', y='Clicks', hue='Channel_Used', data=df)
plt.title("Clicks vs Impressions")
plt.show()
```

## Time-Based Analysis

#### Convert Date Column

```python
df['Date'] = pd.to_datetime(df['Date'])
```

#### Histogram — Campaign Duration

```python
plt.figure(figsize=(8,5))
plt.hist(df['Duration'], bins=20)
plt.title("Distribution of Campaign Duration")
plt.xlabel("Duration")
plt.show()
```

#### Line Chart — Conversion Rate Trend by Company

```python
conv_time = df.groupby(['Date','Company'])['Conversion_Rate'].mean().reset_index()

sns.lineplot(data=conv_time, x='Date', y='Conversion_Rate', hue='Company')
plt.title("Conversion Rate Trend by Company")
plt.show()
```

#### Line Chart — Engagement Score Trend

```python
eng_time = df.groupby('Date')['Engagement_Score'].mean().reset_index()

sns.lineplot(data=eng_time, x='Date', y='Engagement_Score')
plt.title("Engagement Score Trend Over Time")
plt.show()
```

## Geographic Analysis

#### Highest Acquisition Cost by Location

```python
loc_cost = df.groupby('Location')['Acquisition_Cost'].mean().reset_index()

plt.figure(figsize=(10,6))
sns.barplot(x='Location', y='Acquisition_Cost', data=loc_cost)
plt.xticks(rotation=45)
plt.title("Acquisition Cost by Location")
plt.show()
```

#### Conversion Rate by Location and Target Audience

```python
plt.figure(figsize=(10,6))
sns.barplot(x='Location', y='Conversion_Rate', hue='Target_Audience', data=df)
plt.xticks(rotation=45)
plt.title("Conversion Rate by Location and Target Audience")
plt.show()
```

#### Pie Chart — ROI by Location

```python
roi_loc = df.groupby('Location')['ROI'].sum()

plt.figure(figsize=(6,6))
plt.pie(roi_loc, labels=roi_loc.index, autopct='%1.1f%%')
plt.title("ROI Distribution by Location")
plt.show()
```
